## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

1.Hypothesis 1: There is a significant difference in the mean IMDb scores of Movies and TV Shows on Amazon Prime.
2.Hypothesis 2: There is a significant positive relationship between IMDb Score and TMDB Score for Amazon Prime titles.
3.Hypothesis 3: There is a significant difference in the mean runtime of Movies and TV Shows on Amazon Prime.Answer Here.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀): There is no significant difference in the mean IMDb scores of Movies and TV Shows on Amazon Prime.

Alternative Hypothesis (H₁): There is a significant difference in the mean IMDb scores of Movies and TV Shows on Amazon Prime.

#### 2. Perform an appropriate statistical test.

```python
# Perform Statistical Test to obtain P-Value

from scipy.stats import ttest_ind

# Select required columns and remove missing values
hyp1_data = titles[['type', 'imdb_score']].dropna()

# Separate IMDb scores for Movies and TV Shows
movie_scores = hyp1_data[hyp1_data['type'] == 'MOVIE']['imdb_score']
show_scores = hyp1_data[hyp1_data['type'] == 'SHOW']['imdb_score']

# Perform independent two-sample t-test
t_statistic, p_value = ttest_ind(
    movie_scores,
    show_scores,
    equal_var=False
)

print("Mean IMDb Score - Movies:", round(movie_scores.mean(), 2))
print("Mean IMDb Score - TV Shows:", round(show_scores.mean(), 2))
print("T-statistic:", round(t_statistic, 4))
print("P-value:", round(p_value, 4))

# Significance level
alpha = 0.05

if p_value < alpha:
    print("\nConclusion: Reject the Null Hypothesis (H₀).")
    print("There is a statistically significant difference in IMDb scores between Movies and TV Shows.")
else:
    print("\nConclusion: Fail to Reject the Null Hypothesis (H₀).")
    print("There is no statistically significant difference in IMDb scores between Movies and TV Shows.")
```

##### Which statistical test have you done to obtain P-Value?

Independent two-sample t-test (Welch's t-test)

##### Why did you choose the specific statistical test?

The independent two-sample t-test was chosen because we are comparing the mean IMDb scores of two independent groups: Movies and TV Shows. Welch's version of the t-test is used because it does not require the two groups to have equal variances.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀): There is no significant relationship between IMDb Score and TMDB Score for Amazon Prime titles.

Alternative Hypothesis (H₁): There is a significant positive relationship between IMDb Score and TMDB Score for Amazon Prime titles.

#### 2. Perform an appropriate statistical test.

```python
# Perform Statistical Test to obtain P-Value

from scipy.stats import pearsonr

# Select required columns and remove missing values
hyp2_data = titles[['imdb_score', 'tmdb_score']].dropna()

# Perform Pearson correlation test
correlation, p_value = pearsonr(
    hyp2_data['imdb_score'],
    hyp2_data['tmdb_score']
)

print("Pearson Correlation Coefficient:", round(correlation, 4))
print("P-value:", round(p_value, 4))

# Significance level
alpha = 0.05

if p_value < alpha:
    print("\nConclusion: Reject the Null Hypothesis (H₀).")
    print("There is a statistically significant relationship between IMDb Score and TMDB Score.")
else:
    print("\nConclusion: Fail to Reject the Null Hypothesis (H₀).")
    print("There is no statistically significant relationship between IMDb Score and TMDB Score.")
```

##### Which statistical test have you done to obtain P-Value?

Pearson correlation test (Pearson's correlation coefficient).

##### Why did you choose the specific statistical test?

The Pearson correlation test was chosen because we want to measure the strength and direction of the linear relationship between two numerical variables: IMDb Score and TMDB Score.

### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀): There is no significant difference in the mean runtime of Movies and TV Shows on Amazon Prime.

Alternative Hypothesis (H₁): There is a significant difference in the mean runtime of Movies and TV Shows on Amazon Prime.

#### 2. Perform an appropriate statistical test.

```python
# Hypothesis 3: Runtime of Movies vs TV Shows

from scipy.stats import ttest_ind

# Select required columns and remove missing values
hyp3_data = titles[['type', 'runtime']].dropna()

# Separate runtime for Movies and TV Shows
movie_runtime = hyp3_data[hyp3_data['type'] == 'MOVIE']['runtime']
show_runtime = hyp3_data[hyp3_data['type'] == 'SHOW']['runtime']

# Perform independent two-sample t-test
t_statistic, p_value = ttest_ind(
    movie_runtime,
    show_runtime,
    equal_var=False
)

print("Mean Runtime - Movies:", round(movie_runtime.mean(), 2))
print("Mean Runtime - TV Shows:", round(show_runtime.mean(), 2))
print("T-statistic:", round(t_statistic, 4))
print("P-value:", round(p_value, 4))

# Significance level
alpha = 0.05

if p_value < alpha:
    print("\nConclusion: Reject the Null Hypothesis (H₀).")
    print("There is a statistically significant difference in runtime between Movies and TV Shows.")
else:
    print("\nConclusion: Fail to Reject the Null Hypothesis (H₀).")
    print("There is no statistically significant difference in runtime between Movies and TV Shows.")
```

##### Which statistical test have you done to obtain P-Value?

Independent two-sample t-test (Welch's t-test).

##### Why did you choose the specific statistical test?

The independent two-sample t-test was chosen because we are comparing the mean runtime of two independent groups: Movies and TV Shows. Welch's t-test is appropriate because it does not assume that the two groups have equal variances.

## ***6. Feature Engineering & Data Pre-processing***

### 1. Handling Missing Values

```python
# Create dataset for Machine Learning

ml_data = titles[
    [
        'type',
        'release_year',
        'age_certification',
        'runtime',
        'genres',
        'seasons',
        'imdb_votes',
        'tmdb_popularity',
        'tmdb_score',
        'imdb_score'
    ]
].copy()

print("ML Dataset Shape:", ml_data.shape)

display(ml_data.head())
```

```python
# 6.1 Handling Missing Values

# Remove rows where the target variable is missing
ml_data = ml_data.dropna(subset=['imdb_score']).copy()

# Numerical columns
numeric_cols = [
    'release_year',
    'runtime',
    'seasons',
    'imdb_votes',
    'tmdb_popularity',
    'tmdb_score'
]

# Fill numerical missing values with median
for col in numeric_cols:
    ml_data[col] = ml_data[col].fillna(ml_data[col].median())

# Categorical columns
categorical_cols = [
    'type',
    'age_certification',
    'genres'
]

# Fill categorical missing values with mode
for col in categorical_cols:
    ml_data[col] = ml_data[col].fillna(ml_data[col].mode()[0])

# Check missing values
print("Missing Values After Treatment:")
print(ml_data.isnull().sum())
```

#### What all missing value imputation techniques have you used and why did you use those techniques?

Median imputation was used for missing numerical values because it is less sensitive to outliers, while mode imputation was used for categorical variables because it preserves the most common category. Rows with missing imdb_score, the target variable, were removed because they cannot be used for supervised learning. After applying these techniques, all selected features contained zero missing values.

### 2. Handling Outliers

```python
# 6.2 Handling Outliers

# Numerical columns for outlier detection
outlier_cols = [
    'release_year',
    'runtime',
    'seasons',
    'imdb_votes',
    'tmdb_popularity',
    'tmdb_score'
]

# Detect outliers using IQR
for col in outlier_cols:
    Q1 = ml_data[col].quantile(0.25)
    Q3 = ml_data[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = ml_data[
        (ml_data[col] < lower_limit) |
        (ml_data[col] > upper_limit)
    ]

    print(col)
    print("Lower Limit:", round(lower_limit, 2))
    print("Upper Limit:", round(upper_limit, 2))
    print("Number of Outliers:", len(outliers))
    print("-" * 50)
```

##### What all outlier treatment techniques have you used and why did you use those techniques?

The IQR method was used to identify potential outliers in numerical variables such as runtime, seasons, imdb_votes, tmdb_popularity, and tmdb_score. We did not remove the detected outliers because these extreme values may represent valid and meaningful characteristics of Amazon Prime titles. Removing them could result in loss of important information. Therefore, the detected outliers were retained in the dataset for further analysis and machine learning.

### 4. Textual Data Preprocessing
(It's mandatory for textual dataset i.e., NLP, Sentiment Analysis, Text Clustering etc.)

#### 1. Expand Contraction

```python
# Expand contractions in the description column

import re

contractions = {
    "can't": "cannot",
    "won't": "will not",
    "n't": " not",
    "'re": " are",
    "'s": " is",
    "'d": " would",
    "'ll": " will",
    "'t": " not",
    "'ve": " have",
    "'m": " am"
}

def expand_contractions(text):
    if pd.isna(text):
        return text

    text = str(text)

    for contraction, expansion in contractions.items():
        text = re.sub(
            contraction,
            expansion,
            text,
            flags=re.IGNORECASE
        )

    return text

# Apply contraction expansion
titles['description'] = titles['description'].apply(expand_contractions)

print("Contraction expansion completed successfully!")

display(titles[['title', 'description']].head())
```

#### 2. Lower Casing

```python
# Convert description text to lowercase

titles['description'] = titles['description'].apply(
    lambda x: x.lower() if pd.notna(x) else x
)

print("Lower casing completed successfully!")

display(titles[['title', 'description']].head())
```

#### 3. Removing Punctuations

```python
# Remove punctuations from description

import string

def remove_punctuation(text):
    if pd.isna(text):
        return text

    text = str(text)

    for char in string.punctuation:
        text = text.replace(char, '')

    return text

titles['description'] = titles['description'].apply(remove_punctuation)

print("Punctuation removal completed successfully!")

display(titles[['title', 'description']].head())
```

#### 4. Removing URLs & Removing words and digits contain digits.

```python
# Remove URLs and words containing digits

import re

def clean_text(text):
    if pd.isna(text):
        return text

    text = str(text)

    # Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', '', text)

    # Remove words containing digits
    text = re.sub(r'\b\w*\d\w*\b', '', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text

titles['description'] = titles['description'].apply(clean_text)

print("URLs and words containing digits removed successfully!")

display(titles[['title', 'description']].head())
```

#### 5. Removing Stopwords & Removing White spaces

```python
# Remove Stopwords

import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    if pd.isna(text):
        return text

    words = str(text).split()

    filtered_words = []

    for word in words:
        if word not in stop_words:
            filtered_words.append(word)

    return ' '.join(filtered_words)

titles['description'] = titles['description'].apply(remove_stopwords)

print("Stopword removal completed successfully!")

display(titles[['title', 'description']].head())
```

```python
# Remove White Spaces

def remove_whitespaces(text):
    if pd.isna(text):
        return text

    text = str(text)

    # Remove extra spaces between words
    text = ' '.join(text.split())

    return text

titles['description'] = titles['description'].apply(remove_whitespaces)

print("White space removal completed successfully!")

display(titles[['title', 'description']].head())
```

#### 6. Rephrase Text

```python
# Rephrase Text

import random

synonyms = {
    "good": ["great", "excellent"],
    "bad": ["poor", "negative"],
    "story": ["plot", "narrative"],
    "movie": ["film"],
    "shows": ["presents", "features"],
    "about": ["regarding", "concerning"],
    "young": ["youthful"],
    "large": ["big"],
    "small": ["little"],
    "beautiful": ["attractive"]
}

def rephrase_text(text):
    if pd.isna(text):
        return text

    words = str(text).split()
    new_words = []

    for word in words:
        lower_word = word.lower()

        if lower_word in synonyms:
            new_words.append(random.choice(synonyms[lower_word]))
        else:
            new_words.append(word)

    return ' '.join(new_words)

titles['rephrased_description'] = titles['description'].apply(rephrase_text)

print("Text rephrasing completed successfully!")

display(
    titles[['title', 'description', 'rephrased_description']].head()
)
```

#### 7. Tokenization

```python
# Tokenization

import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

def tokenize_text(text):
    if pd.isna(text):
        return []

    return word_tokenize(str(text))

titles['tokenized_description'] = titles['description'].apply(tokenize_text)

print("Tokenization completed successfully!")

display(
    titles[['title', 'description', 'tokenized_description']].head()
)
```

#### 8. Text Normalization

```python
# Normalizing Text using Lemmatization

import nltk
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

def lemmatize_text(tokens):
    if not tokens:
        return []

    lemmatized_words = []

    for word in tokens:
        lemmatized_words.append(lemmatizer.lemmatize(word))

    return lemmatized_words

titles['lemmatized_description'] = titles['tokenized_description'].apply(
    lemmatize_text
)

print("Text normalization using lemmatization completed successfully!")

display(
    titles[
        ['title', 'tokenized_description', 'lemmatized_description']
    ].head()
)
```

##### Which text normalization technique have you used and why?

We used Lemmatization as the text normalization technique. It was chosen because it converts words into their meaningful base or dictionary form while preserving their original meaning. This helps reduce different forms of the same word into a common representation and makes the text more consistent for further NLP processing.

#### 9. Part of speech tagging

```python
# POS Tagging

import nltk
from nltk import pos_tag

nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

def pos_tag_text(tokens):
    if not tokens:
        return []

    return pos_tag(tokens)

titles['pos_tagged_description'] = titles['tokenized_description'].apply(
    pos_tag_text
)

print("POS tagging completed successfully!")

display(
    titles[
        ['title', 'tokenized_description', 'pos_tagged_description']
    ].head()
)
```

```python
# Vectorizing Text using TF-IDF

from sklearn.feature_extraction.text import TfidfVectorizer

# Replace missing descriptions with empty strings
text_data = titles['description'].fillna('')

# Create TF-IDF vectorizer
tfidf = TfidfVectorizer(
    max_features=5000,
    stop_words='english'
)

# Transform text into numerical features
tfidf_matrix = tfidf.fit_transform(text_data)

print("TF-IDF Vectorization completed successfully!")
print("Number of documents:", tfidf_matrix.shape[0])
print("Number of features:", tfidf_matrix.shape[1])
```

##### Which text vectorization technique have you used and why?

We used TF-IDF (Term Frequency–Inverse Document Frequency) for text vectorization. It was chosen because it converts textual descriptions into numerical features based on the importance of each word in a description relative to the entire dataset. TF-IDF gives higher importance to words that are more informative and lower importance to very common words. This makes the text suitable for further machine learning and NLP analysis.

```python
#1. Hyperparameter Optimization + Fit the Algorithm
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import GridSearchCV

# Create Gradient Boosting Regressor
model_3 = GradientBoostingRegressor(random_state=42)

# Define hyperparameter grid
param_grid_3 = {
    'n_estimators': [50, 100],
    'learning_rate': [0.05, 0.1],
    'max_depth': [2, 3],
    'min_samples_split': [2, 5]
}

# Apply GridSearchCV
grid_search_3 = GridSearchCV(
    estimator=model_3,
    param_grid=param_grid_3,
    cv=5,
    scoring='neg_mean_squared_error',
    n_jobs=-1
)

# Fit the model
grid_search_3.fit(X_train, y_train)

# Get the best model
model_3_best = grid_search_3.best_estimator_

print("Best Parameters:")
print(grid_search_3.best_params_)

print("\nBest Gradient Boosting model trained successfully!")
```
